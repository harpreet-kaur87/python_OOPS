#### 🚀 Mini Project: Bank System (OOP - Python)

🧠 Core Idea  
Build a simple banking system using Object-Oriented Programming where multiple users can manage their bank accounts.  

🧩 Required Features  

1. 🏦 BankAccount class  
Private __balance  
Methods:  
deposit(amount)  
withdraw(amount)  
get_balance()  

2. 👤 User class  
name  
account (BankAccount object)  
account_number (unique for each user)  

3. 🏢 Bank class  
Private list of users  
Methods:  
create_account(name, initial_balance)  
find_user(account_number)  
deposit_to_user(account_number, amount)  
withdraw_from_user(account_number, amount)  
show_balance(account_number)  

🎯 Functionality  
Create account with unique account number  
Deposit money  
Withdraw money (with validation)  
Check account balance  

```python
class BankAccount:
    
    def __init__ (self,balance):
        self.__balance = balance
        
    def deposit(self,amount):
        if amount <= 0:
            return f'Enter valid amount'
        else:
            self.__balance += amount
            return f'Current Balance : {self.__balance}'
        
    def withdraw(self,amount):
        if amount <= 0:
            return f'Enter valid amount'
        elif amount > self.__balance:
            return f'Insufficient balance'
        else:
            self.__balance -= amount
            return f'Current Balance : {self.__balance}'
        
    def get_balance(self):
        return self.__balance
        
class User:
    
    def __init__ (self,name,account,account_number):
        self.name = name
        self.account = account
        self.account_number = account_number
        
class Bank:
    
    def __init__ (self):
        self.__userList = []
        self.__account_number = 1001
        
    def create_account(self,name,initial_balance):
        acc = BankAccount(initial_balance)
        account_number = self.__account_number
        self.__account_number += 1
        user = User(name,acc,account_number)
        self.__userList.append(user)
        return account_number
        
    def find_user(self,account_number):
        for user in self.__userList:
            if user.account_number == account_number:
                return user
        return None
    
    def deposit_to_user(self,account_number,amount):
        user = self.find_user(account_number)
        
        if user:
            return user.account.deposit(amount)
        else:
            return f'user not found'
        
    def withdraw_from_user(self,account_number,amount):
        user = self.find_user(account_number)
        
        if user:
            return user.account.withdraw(amount)
        else:
            return f'user not found'
        
    def show_balance(self,account_number):
        user = self.find_user(account_number)
        
        if user:
            return user.account.get_balance()
        else:
            return f'user not found'
        
bank = Bank()
acc1 = bank.create_account('Damini',200000)
print(acc1)
acc2 = bank.create_account('Namit',50000)
print(acc2)
print(bank.deposit_to_user(acc1,25000))
print(bank.show_balance(acc1))
print(bank.withdraw_from_user(acc2,20000))
acc3 = bank.create_account('Harpreet',65000)
print(acc3)
print(bank.show_balance(acc3))
```
